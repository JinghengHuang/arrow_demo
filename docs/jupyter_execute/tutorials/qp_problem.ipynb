# LP Example: Production Planning Optimization

In this example, we solve a **linear programming (LP)** problem to optimize a small factory's production plan.

## Problem Description

Consider a small manufacturing workshop that produces two products: **Product A** and **Product B**. Each day, the workshop operates under limited resources—specifically, labor hours and raw materials. The objective is to determine an optimal production plan that **maximizes profit** while respecting these resource constraints.

**Product A** yields a profit of \$40 per unit and consumes more labor.  
**Product B** yields a profit of \$30 per unit and consumes more raw materials.

The available daily resources are:
- **100 units of labor**
- **80 units of raw material**

The problem is to determine how many units of Product A and Product B should be produced in order to achieve the highest profit without exceeding resource limits.

Let’s model this as a **linear programming** problem.

## Mathematical Formulation

Let:
- \( x \) be the number of units of Product A to produce  
- \( y \) be the number of units of Product B to produce

### Objective:
Maximize total profit:  
$$
\text{maximize } 40x + 30y
$$

### Constraints:
- Labor usage: \( 2x + y \leq 100 \)  
- Material usage: \( x + 2y \leq 80 \)  
- Non-negativity: \( x, y \geq 0 \)




## Solving with OptArrow

The following code illustrates how this linear program can be defined and solved using **OptArrow** in Python:


```python
import pyarrow as pa
from optarrow.client import OptArrowClient
from optarrow.model import LPModel

model = LPModel(
    c = [40.0, 30.0],         # Profit coefficients
    A = [[2.0, 1.0],          # Labor constraint
         [1.0, 2.0]],         # Material constraint
    b = [100.0, 80.0],        # Resource limits
    lb = [0.0, 0.0],          # Variables must be non-negative
    osense = "max"            # Maximize the objective
)
```
